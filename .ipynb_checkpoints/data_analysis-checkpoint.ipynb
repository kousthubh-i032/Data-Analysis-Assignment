#### 3. Handling missing values : 

#### 3a. Removal of unnecessary columns : 

```python
# 'Unnamed: 22', 'promotion-ids', 'fulfilled-by' columns are of no use, so they are being dropped.
newdf = df.drop(columns=['Unnamed: 22', 'fulfilled-by'])
```

#### 3b. Missing values checking : 

```python
(newdf.isna().mean() * 100).sort_values(ascending=False)
```

```python
newdf[newdf['Amount'].isna()]
```

```python
newdf['Amount']=df['Amount'].fillna(0.00)
```

```python
newdf['currency']=df['currency'].fillna('INR')
```

```python
(newdf.isna().mean() * 100).sort_values(ascending=False)
```

```python
newdf=newdf.drop(columns=['index','ship-country'])
```

```python
newdf['promotion-ids']=newdf['promotion-ids'].fillna('no')
```

```python
newdf.loc[(newdf['Courier Status'].isna()) & (newdf['Status']=='Cancelled'),'Courier Status']='Cancelled'
```

```python
newdf['Courier Status']=newdf['Courier Status'].fillna('Cancelled')
```

```python
(newdf.isna().mean() * 100).sort_values(ascending=False)
```

#### 4. Converted date fields to proper datetime format.

```python
df['Date'].dropna().unique()
```

```python
newdf['Date']=pd.to_datetime(df['Date'],format='%m-%d-%y',errors='coerce')
```
