#### 4. Converted date fields to proper datetime format.

#### 5. Standardize categorical variables

```python
newdf.columns
```

```python
for col in newdf.columns:
    print(f"\n{col}:")
    print(newdf[col].unique())
```

#### Column 'Fulfilment','ship-service-level','Courier Status','Sales Channel '   values will be lowercased

```python
newdf['Fulfilment']=newdf['Fulfilment'].str.lower()
newdf['ship-service-level']=newdf['ship-service-level'].str.lower()
newdf['Courier Status']=newdf['Courier Status'].str.lower()
newdf['Sales Channel ']=newdf['Sales Channel '].str.lower()
```

#### Removal of " Dress" from the 'Category' column

```python
newdf['Category']=newdf['Category'].str.replace(' Dress','',regex=False)
newdf['Category']=newdf['Category'].str.lower()
```

#### Handling 'ship-state' column 

```python
newdf['ship-state']=newdf['ship-state'].str.lower().str.strip()
```

```python
states_correction = {
    'pb': 'punjab',
    'rj': 'rajasthan',
    'nl': 'nagaland',
    'ar': 'arunachal pradesh',
    'apo': 'andhra pradesh',
    'rajshthan': 'rajasthan',
    'rajsthan': 'rajasthan',
    'new delhi': 'delhi',
    'punjab/mohali/zirakpur': 'punjab',
    'orissa': 'odisha',
    'pondicherry': 'puducherry',
    'andaman & nicobar': 'andaman & nicobar islands'
}
```

```python
newdf['ship-state']=newdf['ship-state'].replace(states_correction)
```

```python
newdf['ship-state']=newdf['ship-state'].str.title()
```

#### New address_df creation for future reference

```python
address_df=newdf[['ship-city','ship-state','ship-postal-code']].copy()
```

#### 6. Creation of New Useful Columns from the existing ones

Creation of 'Month_Name' and 'Day_Name'

```python
newdf['Month_Name']=newdf['Date'].dt.month_name()
newdf['Day_name']=newdf['Date'].dt.day_name()
```

#### Creation of Financial Quarter with year : 

```python
#Standard Quarter system maps Jan, Feb, Mar to Q1
newdf['Standard_Quarter'] = newdf['Date'].dt.quarter

#Mapping Apr, May, June to Q1 for the Indian Financial Year
newdf['Financial_Quarter'] = newdf['Date'].dt.month.map({
    4: 'Q1', 5: 'Q1', 6: 'Q1',
    7: 'Q2', 8: 'Q2', 9: 'Q2',
    10: 'Q3', 11: 'Q3', 12: 'Q3',
    1: 'Q4', 2: 'Q4', 3: 'Q4'
})

#Quarter Year : yyyy-Q# format
newdf['Quarter_Year'] = newdf['Date'].dt.year.astype(str) + '-' + newdf['Financial_Quarter']

newdf.drop(columns=['Standard_Quarter','Financial_Quarter'],inplace=True)

newdf['Quarter_Year'].head()
```

```python
for col in newdf.columns:
    print(f"\n{col}:")
    print(newdf[col].unique())
```

#### Checking if the 'Amount' column is equal to the unit price of the product.

```python
newdf['Revenue']=newdf['Qty']*newdf['Amount']
newdf['Check_Diff']=newdf['Revenue']-newdf['Amount']
```

```python
newdf['Check_Diff'].unique()
```

#### Bulk Buys in the list, where the difference between the Revenue and the Amount is more than 5000

```python
bulk_orders = newdf[newdf['Check_Diff'] > 5000]
print(bulk_orders[['Order ID', 'Category', 'Qty', 'Amount', 'Revenue']])
```

#### Fixing the Revenue Logic :<br> 
#### As we saw that "Amount" is basically unit product price : we will rename the column into "Unit Price"

```python
newdf.rename(columns={'Amount':'Unit_Price'},inplace=True)
```

#### We create final revenue column : 'Total_Revenue'

```python
newdf['Total_Revenue']=newdf['Qty']*newdf['Unit_Price']
```

#### Drop the unnecessary columns : 

```python
newdf.drop(columns=['Revenue','Check_Diff'],inplace=True)
```
